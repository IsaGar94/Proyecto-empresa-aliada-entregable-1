```python
#no hay datos duplicado ya no se elimina, ya no se procede a realizar más código, se investiga y se comenta que con el drop_duplicates() se elimina
#Unir los DataFrames relevantes:Realiza uniones entre los DataFrames para consolidar la información. 
#Por ejemplo, une la tabla de productos con las ventas, las categorías y los segmentos.

df_union1 = a_fact_sales.merge(a_dim_product, left_on='ITEM_CODE', right_on='ITEM', how='left')
df_union1.head()
```

```python
df_union2 = df_union1.merge(a_dim_category,left_on='CATEGORY',right_on='ID_CATEGORY', how='left')
df_union2.head()
```

```python
df_union3 = df_union2.merge( a_dim_segment,
    left_on=['CATEGORY_x', 'ATTR1', 'ATTR2', 'ATTR3', 'FORMAT'],
    right_on=['CATEGORY', 'ATTR1', 'ATTR2', 'ATTR3', 'FORMAT'],
    how='left')
df_union3.head()
```

```python
df_union4 = df_union3.merge(a_dim_calendar,on='WEEK',how='left')
df_union4.head()
```

```python
df_union4.isnull().sum()
```

```python
#Verificar que fecha este bien almacenada
df_union4['DATE'] = pd.to_datetime(df_union4['DATE'])
df_union4['DATE'].dtype
```

```python
#Verificar las columna y ver si hay repetidas y como las manda 
df_union4.columns.tolist()
```

```python
#Limpiar los espacion inecesarios
for columna in df_union4.select_dtypes(include='str').columns:df_union4[columna] = df_union4[columna].str.strip()
df_union4.head()
```

```python
#que no haya espacion inecesarios
df_union4['CATEGORY_y'].unique()
```

```python
# Se recomienda crear una copia por si las operaciones no las realice bien, entonces solo realizare una suma y media
df_union4_resumen = df_union4.copy()

# Calcular resumen usando los datos originales
df_union4_resumen.loc['RESUMEN'] = pd.NA

df_union4_resumen.loc['RESUMEN', 'TOTAL_UNIT_SALES'] = (df_union4['TOTAL_UNIT_SALES'].sum())
df_union4_resumen.loc['RESUMEN', 'TOTAL_VALUE_SALES'] = (df_union4['TOTAL_VALUE_SALES'].sum())
df_union4_resumen.loc['RESUMEN', 'TOTAL_UNIT_AVG_WEEKLY_SALES'] = (df_union4['TOTAL_UNIT_AVG_WEEKLY_SALES'].mean())
df_union4_resumen.tail()
```
